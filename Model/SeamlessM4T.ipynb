```python
import torchaudio
import ipywidgets as widgets
from IPython.display import display, Audio
from transformers import AutoProcessor, SeamlessM4Tv2Model
```

```python
# Chargement du modèle SeamlessM4T
processor = AutoProcessor.from_pretrained("facebook/seamless-m4t-v2-large")
model = SeamlessM4Tv2Model.from_pretrained("facebook/seamless-m4t-v2-large")
```

```python
# Widgets
src_dropdown = widgets.Dropdown(options=sorted(lang_dict.keys()), description='Langue source :', value="English")
tgt_dropdown = widgets.Dropdown(options=sorted(lang_dict.keys()), description='Langue cible :', value="French")

text_input = widgets.Textarea(
    value="Hello, my dog is cute",
    placeholder='Entrez votre texte ici...',
    description='Texte source :',
    layout=widgets.Layout(width='100%', height='100px')
)


mode_dropdown = widgets.Dropdown(
    options=["Texte → Texte", "Texte → Audio"],
    value="Texte → Texte", description="Mode de traduction :"
)

button = widgets.Button(description="Traduire", button_style='success')
output = widgets.Output()
```

```python
# Affichage des widgets
display(mode_dropdown, src_dropdown, tgt_dropdown, text_input,button, output)
```

```python
# Fonction principale
def on_translate_clicked(b):
    output.clear_output()
    src_code = lang_dict.get(src_dropdown.value)
    tgt_code = lang_dict.get(tgt_dropdown.value)
    mode = mode_dropdown.value

    with output:
        try:
            # Mode 1 : Texte → Texte
            if mode == "Texte → Texte":
                text = text_input.value.strip()
                if not text:
                    print("⚠️ Entrez un texte.")
                    return
                inputs = processor(text=text, src_lang=src_code, return_tensors="pt")
                generated = model.generate(**inputs, tgt_lang=tgt_code, generate_speech=False)
                result = processor.decode(generated[0].tolist()[0], skip_special_tokens=True)
                print(f"✅ Traduction {src_dropdown.value} → {tgt_dropdown.value} :\n{result}")

            # Mode 2 : Texte → Audio
            elif mode == "Texte → Audio":
                text = text_input.value.strip()
                if not text:
                    print("⚠️ Entrez un texte.")
                    return
                inputs = processor(text=text, src_lang=src_code, return_tensors="pt")
                audio_array = model.generate(**inputs, tgt_lang=tgt_code)[0].cpu().numpy().squeeze()
                print(f"✅ Audio généré ({src_dropdown.value} → {tgt_dropdown.value}) :")
                display(Audio(audio_array, rate=16000))

        except Exception as e:
            print("❌ Erreur :", str(e))

# Lien bouton → fonction
button.on_click(on_translate_clicked)
```
