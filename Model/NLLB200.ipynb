```python
# Cell 2 : Importations
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, pipeline
from IPython.display import display
import ipywidgets as widgets
```

```python
from transformers.utils import logging
logging.set_verbosity_error()
import sys
import os
import contextlib

@contextlib.contextmanager
def suppress_stdout():
    with open(os.devnull, 'w') as fnull:
        old_stdout = sys.stdout
        sys.stdout = fnull
        try:
            yield
        finally:
            sys.stdout = old_stdout

with suppress_stdout():
  # Nom du modèle sur Hugging Face
  model_name = "facebook/nllb-200-distilled-600M"

  # Chemin local où tu veux le sauvegarder
  save_path = "./nllb_local_model"

  # Télécharger depuis Hugging Face
  tokenizer = AutoTokenizer.from_pretrained(model_name)
  model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

  # Sauvegarder localement
  tokenizer.save_pretrained(save_path)
  model.save_pretrained(save_path)

  print(f"✅ Modèle téléchargé et sauvegardé dans : {save_path}")
```

```python
#  Interface utilisateur

# Menu déroulant pour la langue source
src_dropdown = widgets.Dropdown(
    options=list(lang_dict.keys()),
    description='De :',
    value="Bambara",
    style={'description_width': 'initial'}
)

# Menu déroulant pour la langue cible
tgt_dropdown = widgets.Dropdown(
    options=list(lang_dict.keys()),
    description='Vers :',
    value="French",
    style={'description_width': 'initial'}
)

# Zone de saisie du texte à traduire
text_input = widgets.Textarea(
    value="N muso tε karamɔgɔ ye",
    placeholder='Entrez votre texte ici...',
    description='Texte :',
    layout=widgets.Layout(width='100%', height='100px'),
    style={'description_width': 'initial'}
)

# Bouton de lancement de la traduction
button = widgets.Button(description="Traduire", button_style='success')

# Zone d'affichage du résultat
output = widgets.Output()

# Affichage des widgets dans le notebook
display(src_dropdown, tgt_dropdown, text_input, button, output)
```

```python
#Fonction de traduction
def on_translate_clicked(b):
    output.clear_output()
    src_lang = lang_dict[src_dropdown.value]
    tgt_lang = lang_dict[tgt_dropdown.value]
    text = text_input.value.strip()

    if not text:
        with output:
            print("⚠️ Entrez un texte à traduire.")
        return

    try:
        translator = pipeline(
            'translation',
            model=model,
            tokenizer=tokenizer,
            src_lang=src_lang,
            tgt_lang=tgt_lang,
            max_length=400
        )

        result = translator(text)
        translation = result[0]['translation_text']

        with output:
            print(f"Traduction ({src_dropdown.value} → {tgt_dropdown.value}) :\n")
            print(translation)

    except Exception as e:
        with output:
            print("❌ Erreur lors de la traduction :", e)

button.on_click(on_translate_clicked)
```
